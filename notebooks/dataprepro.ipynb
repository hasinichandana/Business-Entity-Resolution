```python
from google.colab import drive
import os
import shutil

# Mount Google Drive
drive.mount('/content/drive')

drive_path = "/content/drive/MyDrive/ML/student_resource"
local_path = "/content/student_resource"

# Copy to local runtime for fast read/write
if os.path.exists(drive_path):
    if not os.path.exists(local_path):
        print("Copying student_resource to local disk...")
        shutil.copytree(drive_path, local_path)
        print("Copy complete!")
    else:
        print("Folder already present locally.")
        
    dataset_dir = os.path.join(local_path, "dataset")
    print("\nTrain files:", os.listdir(os.path.join(dataset_dir, "train")))
    print("Test files:", os.listdir(os.path.join(dataset_dir, "test")))
else:
    print(f"Directory not found: {drive_path}")
    print("Contents of /content/drive/MyDrive/ml:", os.listdir("/content/drive/MyDrive/ml") if os.path.exists("/content/drive/MyDrive/ml") else "ml folder not found")
```
